# 📉 Notebook 02: Modelo Predictivo de Regresión Lineal y Evaluación de IA (Challenge 2 - UPCH)
* **Objetivo:** Implementación de un algoritmo de Machine Learning (Regresión Lineal) para la predicción de estancia hospitalaria.
* **Enfoque:** Métricas de rendimiento del modelo y análisis de correlación clínica (Sin duplicar gráficos exploratorios).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

print("🤖 [IA Predictiva] Configurando entorno de Machine Learning...")
```

## 🤖 2. Entrenamiento y Validación del Algoritmo de IA
Entrenamos el modelo supervisado para predecir los días de hospitalización en función de la severidad de las patologías respiratorias del MINSA.

```python
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([2.5, 6.0, 10.0, 3.0, 14.0, 8.5])
modelo_ia = LinearRegression()
modelo_ia.fit(X, y)
y_pred = modelo_ia.predict(X)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f" Métricas de Rendimiento del Algoritmo:")
print(f"   - Error Cuadrático Medio (MSE): {mse:.4f}")
print(f"   - Coeficiente de Determinación (R²): {r2:.4f}")

severidad_neumonia = np.array([[5]])
estancia_predicha = modelo_ia.predict(severidad_neumonia)

print(f"\n [Resultado de la Predicción]")
print(f"   - Patología: Influenza y Neumonía (J09-J18)")
print(f"   - Estancia hospitalaria estimada: ** {estancia_predicha[0]:.1f} días **")
```

## 📈 3. Visualización Exclusiva del Modelo Predictivo (Regresión y Residuos)
Generamos gráficos especializados en Machine Learning: la curva de ajuste de la regresión con el punto de predicción y un gráfico de residuos del modelo para validar el error.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfica 1: Curva de Regresión Lineal y Predicción de Neumonía
ax1.scatter(X, y, color='#1f77b4', s=90, label='Datos Históricos (Entrenamiento)')
ax1.plot(X, y_pred, color='#d62728', linestyle='--', linewidth=2, label='Línea de Tendencia (Regresión IA)')
ax1.scatter(severidad_neumonia, estancia_predicha, color='#2ca02c', s=220, edgecolor='black', zorder=5, label='Predicción Neumonía (J09-J18)')
ax1.set_title('Modelo Predictivo: Severidad vs Estancia Hospitalaria', fontsize=11, fontweight='bold')
ax1.set_xlabel('Índice de Severidad Clínica', fontsize=10)
ax1.set_ylabel('Días de Estancia', fontsize=10)
ax1.legend(loc='upper left', fontsize=9)
ax1.grid(True, linestyle=':', alpha=0.7)
residuos = y - y_pred
ax2.axhline(0, color='black', linestyle='-', linewidth=1)
ax2.scatter(y_pred, residuos, color='#ff7f0e', s=90, edgecolor='black')
ax2.set_title('Gráfico de Residuos (Evaluación de Errores del Modelo)', fontsize=11, fontweight='bold')
ax2.set_xlabel('Valores Predichos', fontsize=10)
ax2.set_ylabel('Residuos (Real - Predicho)', fontsize=10)
ax2.grid(True, linestyle=':', alpha=0.7)

plt.tight_layout()
plt.show()
```
